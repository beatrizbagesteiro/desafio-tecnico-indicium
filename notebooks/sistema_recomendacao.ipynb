```python
import pandas as pd
import numpy as np
```

```python
path = '../data/raw/'
raw_products_data = pd.read_csv(f"{path}products.csv")
raw_product_variants_data =  pd.read_csv(f"{path}product_variants.csv")
raw_orders_data = pd.read_csv(f"{path}orders.csv")
raw_orders_items_data = pd.read_csv(f"{path}order_items.csv")

products = raw_products_data.copy(deep=True)
products_variants = raw_product_variants_data.copy(deep=True)
orders = raw_orders_data.copy(deep=True)
order_items = raw_orders_items_data.copy(deep=True)
```

```python
products
```

```python
produto_alvo = products.loc[
    (products["name"] == "Motor de Popa 1949"), 
    ['id']
].copy()
produto_alvo
```

```python
products_variants
```

```python
products_active = products[products['is_active'] == 'active']
df_products = pd.merge(products, products_variants, left_on='id', right_on='product_id', how='inner')
df_products = df_products[['id_x','id_y']].copy()
df_products = df_products.rename(columns={'id_x':'product_id','id_y':'variant_id'})
df_products
```

```python
order_items
```

```python
df_orders = pd.merge(df_products,order_items, left_on='variant_id', right_on='product_variant_id')
df_orders = df_orders[['product_id', 'order_id']].copy()
df_orders
```

```python
orders['placed_at'] = pd.to_datetime(orders['placed_at'])
orders_paid = orders[(orders['status']=='paid') & (orders['placed_at'] <= pd.Timestamp.today())].copy()
df_orders = pd.merge(df_orders,orders_paid, left_on='order_id', right_on='id')
df_orders = df_orders[['product_id', 'customer_id']].copy()
df_orders
```

```python
matriz_usuario_produto = pd.crosstab(index=df_orders['customer_id'], columns=df_orders['product_id'])
array_usuario_produto = np.where(matriz_usuario_produto > 0, 1, 0)
array_usuario_produto
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
'''
Similaridade de Cosseno calcula as linhas, como queremos o cálculo entre produtos tem que transpor a matriz (colocar os produtos na linha)
'''
similaridade_produtos = cosine_similarity(matriz_usuario_produto.T)

similaridade_produtos = pd.DataFrame(similaridade_produtos, matriz_usuario_produto.columns, matriz_usuario_produto.columns)
similaridade_produtos
```

```python
id_motor_popa_1949 = produto_alvo['id'].item()
top_5 = similaridade_produtos[id_motor_popa_1949].sort_values(ascending=False)
top_5 = top_5.drop(id_motor_popa_1949)
top_5 = top_5.head(5)
top_5 = top_5.reset_index()
```

```python

ranking = pd.merge(top_5, raw_products_data[['id', 'name']], left_on='product_id', right_on='id')

ranking = ranking.rename(columns={180:'cosine_similarity', 'name':'product_name'})
ranking = ranking.drop(columns=['id'])
ranking = ranking[['product_id', 'product_name', 'cosine_similarity']]

ranking
```
